### My solution

```python
# start your code here
x = np.arange(-10, 10)
y2 = -0.9 + x
y3 = 8.9 - x
y = y2 * ((0.9 <= x) & (x < 4.9)) + y3 * ((4.9 <= x) & (x < 8.9))

plt.figure(figsize = [10, 5], facecolor = color3)
plt.tick_params(labelcolor = "w")
plt.title("System f(x)", color = "w")
plt.xlabel("x", color = "w")
plt.ylabel("y", color = "w")
plt.plot(x, y, "d-", color = color2, label = "f(x)")
plt.grid(True)
plt.legend(loc = "upper right")
```
