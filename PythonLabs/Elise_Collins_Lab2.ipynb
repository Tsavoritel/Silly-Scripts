```python
# Your code starts here
x = np.arange(0, 10, 0.01)
y1 = np.sqrt(3.4 * x)
y2 = np.pow(x - 4.51, 2)

fig, ax = plt.subplots(3, figsize = (10, 10), sharex = True, sharey = True)

plt.sca(ax[0])
plt.plot(x, y1, color=color1, label="$f(x) = \\sqrt{3.4 \\bullet x}$")
plt.xlabel("x")
plt.ylabel("y")
plt.title("f(x)")
plt.grid()
plt.legend()

plt.sca(ax[1])
plt.plot(x, y2, color=color2, label="$g(x) = (3.4 \\bullet x)^2$")
plt.xlabel("x")
plt.ylabel("y")
plt.title("g(x)")
plt.grid()
plt.legend()

plt.sca(ax[2])
plt.plot(x, y1, color=color1, label="$f(x) = \\sqrt{3.4 \\bullet x}$")
plt.plot(x, y2, color=color2, label="$g(x) = (3.4 \\bullet x)^2$")
plt.xlabel("x")
plt.ylabel("y")
plt.title("f(x) and g(x)")
plt.grid()
plt.legend()
```
