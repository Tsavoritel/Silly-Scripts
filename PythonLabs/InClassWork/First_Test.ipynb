```python
x = np.arange(0, 4, 0.05)
y = np.sin(x * np.pi)

#Object Oriented Approach
fig, ax = plt.subplots(figsize = (6, 4)) # figure and axis, subplot? in figure
ax.plot(x, y)
ax.set_xlabel("Time [s]")
ax.set_ylabel("Amplitude [V]")
ax.set_title("Sine wave")
fig.set_facecolor("#8060a0")
```

```python
# Pyplot Approach
x = np.arange(0, 4, 0.05)
y = np.sin(x * np.pi)

plt.figure(figsize = [8, 5], facecolor = "#4e7241")
plt.plot(x, y, color = "#4e7241")
plt.xlabel("Time [s]")
plt.ylabel("Amplitude [V]")
plt.title("Sine wave")
plt.grid(True)
plt.axhline(0.0, color = "#a01090")
plt.axvline(2.0, color = "#a01090")
```

```python
x = np.linspace (-5, 5, 200)
y = x**2 - 2*x - 5
plt.figure(figsize=(8, 5))
plt.plot(x, y, label="f(x)")
plt.axhline(0, color="red", linestyle="--", label="$y=0$")
plt.title("Graphical solution of the equation $f(x) = 0$")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.xlim(-1.5,-1.4)
plt.ylim(-0.5, 0.5)
plt.scatter(-1.45, 0, color="r")
plt.text(-1.44, 0.05, "$x \\approx - 1.45 $", color="red")
plt.legend()
plt.grid(True)
plt.show()
```
