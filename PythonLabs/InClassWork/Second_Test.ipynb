```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
1.0 * (x < 4) # when x < 4 then multiply that mask by 1
```

```python
# piecewise function
x * (x < 4) + x**2 * (x >= 4)
```

```python
#calc x when 3 < x <= 8
x * ((3 < x) & (x < 8))
```

```python
def step (x):
    return 1.0*(x>=0)
(step(x-3) - step(x-6))*x/2
((x >= 3) & (x<6)) * x/2
```

```python
x = np.arange(-5,5, 0.01)
y = 0.25 * x ** 2
b = y > 2

plt.subplots(figsize = (10, 4), sharex = True, sharey = True)

plt.plot(x, y, color = "#aa555588")
plt.grid(True)

plt.plot(x, b, color = "#55aa5588")
plt.grid(True)

plt.plot(x, b * y, color = "#5555aa88")
plt.grid(True)

plt.show()
```

```python
x = np.arange(-5, 5, 1)
y1 = x**2 - 10
y2 = x * 4
y = (x < 0 ) * y1 + (x >= 0) * y2

plt.subplots(figsize = (10, 4), sharex = True, sharey = True)

plt.plot(x, y1, 'd-', color = "#aa555588", label = "y1")
plt.grid(True)
plt.legend(loc = "lower right")

plt.plot(x, y2, 's-', color = "#55aa5588", label = "y2")
plt.grid(True)
plt.legend(loc = "lower right")

plt.plot(x, y, 'o-', color = "#5555aa88", label = "y")
plt.grid(True)
plt.legend(loc = "lower right")

plt.show()
```
